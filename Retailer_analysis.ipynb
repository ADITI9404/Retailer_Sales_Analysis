```python
df=df.drop_duplicates()
```

```python
df['Transaction_ID'].duplicated().sum()
```

```python
df.drop_duplicates(subset='Transaction_ID',inplace=True)
```

```python
df.dropna(subset=['Transaction_ID'], inplace=True)
```

```python
df.dropna(subset=['Customer_ID'], inplace=True)
```

```python
df['Ratings']=df.groupby('Product_Type')['Ratings'].transform(lambda x: x.fillna(x.median()))
```

```python
df['Time']=df['Time'].fillna(df['Time'].mode()[0])
```

```python
df['Date']=df['Date'].fillna(method='ffill')
```

```python
df['Date']=df['Date'].str.replace('-','/')
```

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
df['Year']=df['Date'].dt.year
```

```python
df['Month']=df['Date'].dt.month_name()
```

```python
df['Gender']=df.groupby('Product_Category')['Gender'].transform(lambda x:x.fillna(x.mode()))
```

```python
df['Name']=df['Name'].fillna('Unknown')
```

```python
df['Email']=df['Email'].fillna('Not available')
```

```python
df['Address']=df['Address'].fillna('Unknown')
```

```python
df['City']=df['City'].fillna(df['City'].mode())
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df['Phone']=df['Phone'].fillna('Not Available')
```

```python
df['State']=df['State'].fillna(df['State'].mode()[0])
```

```python
df['Zipcode']=df['Zipcode'].fillna(-1)
```

```python
df['Country']=df['Country'].fillna(df['Country'].mode()[0])
```

```python
df['Customer_Segment']=df['Customer_Segment'].fillna('Unknown')
```

```python
df['Product_Category']=df['Product_Category'].fillna('Unknown')
```

```python
df['Feedback']=df['Feedback'].fillna('Unknown')
```

```python
df['Product_Brand']=df['Product_Brand'].fillna('Unknown')
```

```python
df['Shipping_Method']=df['Shipping_Method'].fillna('Unknown')
```

```python
df['Payment_Method']=df['Payment_Method'].fillna('Unknown')
```

```python
df['Order_Status']=df['Order_Status'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df[['Transaction_ID', 'Customer_ID','Zipcode','Age','Year','Total_Purchases']]=df[['Transaction_ID', 'Customer_ID','Zipcode','Age','Year','Total_Purchases']].astype('int64')
```

```python
df['Age_Group']=pd.qcut(df['Age'],q=4,labels=['Young','Adult','Middle Age','Senior'])
```

```python
hour = pd.to_timedelta(df['Time']).dt.components['hours']
bins=[0,4,12,16,20,24]
labels=['midnight','morning','afternoon','evening','night']
df['time_period']=pd.cut(hour,bins=bins,labels=labels)
```
